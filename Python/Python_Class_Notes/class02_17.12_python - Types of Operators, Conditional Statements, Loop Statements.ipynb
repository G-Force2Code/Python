### Bitwise Operators

```python
# & Bitwise AND
# | Bitwise OR
# ^ Bitwise XOR
# XOR Operation says that the o/p will be true iff both the i/ps are different, otherwise False
```

```python
11 & 5

# 11 --> 1 0 1 1
#  5 --> 0 1 0 1
# ---------------
#        0 0 0 1 --> 1
# ---------------
```

```python
11 | 5

# 11 --> 1 0 1 1
#  5 --> 0 1 0 1
# ---------------
#        1 1 1 1  --> 15
# ---------------
```

```python
11 ^ 5

# 11 --> 1 0 1 1
#  5 --> 0 1 0 1
# ---------------
#        1 1 1 0  --> 14
# ---------------
```

```python
# ladder if
marks = int(input('Enter marks: '))
if marks >= 90:
    print('Grade A')
if marks >= 80 and marks < 90:
    print('Grade B')
if marks >= 70 and marks < 80:
    print('Grade C')
else:
    print('Failed')
```

```python
# marks = int(input('Enter marks: '))
# if marks > 40:
#     print('Passed')
# else:
#     print('Failed')
    
# Shorthand Notation
marks = int(input('Enter marks: '))
print('Passed') if marks > 40 else print("Failed")
```

```python
marks = int(input('Enter marks: '))
if marks >= 90:
    print('Grade A')
else:
    if marks >= 80:
        print('Grade B')
    else:
        print('Grade C')
        
print('Grade A') if marks > 90 else print('Grade B') if marks > 80 else print('Grade C')
```

#### range function
It creates a sequence of integers in a given range with given StepSize.

```python
# range(startValue, endValue, stepSize = 1)
range(n):     0, 1, 2................(n-1)
range(m, n):  m, m+1, m+2, ..........(n-1)
range(m,n,p): m, m+p, m+2p...........(n-1)
```

```python
# flag concept # Here status is being used as a flag variable
status = True                                   
while(status):
    marks = int(input('Enter marks: '))
    if marks < 0:
        status =  False
    elif marks > 40:
        print('Passed')
    else:
        print('Failed')
```
