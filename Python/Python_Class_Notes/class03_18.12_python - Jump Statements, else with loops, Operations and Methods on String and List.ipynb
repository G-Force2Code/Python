### continue statement
Whenever loop encounters a continue statement, it will terminate the current iteration and the control will pass on to the next iteration

```python
for x in range(1, 10):
    print(x)
```

```python
for x in range(1, 10):
    if x == 5:
        break
    print(x)
```

```python
for x in range(1, 10):
    if x == 5:
        continue
    print(x)
```

```python
for x in range(1, 10):
    if x < 5:
        continue
    print(x)
```

```python
for x in range(1, 10):
    if x > 7:
        continue
    print(x)
```

```python
for x in range(1, 10):
    print(x)
    if x > 5:
        continue
```

```python
marks = int(input('Enter marks: '))
if marks > 40:
    pass
else:
    print('Failed')
```

### else statement in loop statements
else block parallel to loop statement will only get executed when loop completes all the iterations.
In other words, if the loop terminates without running through all iterations, else block won't get executed.

```python
for x in range(10):
    if x == 5:
        break
    print(x)
```

```python
for x in range(10):
    if x == 5:
        break
    print(x)
else:
    print('Thanks')
```

```python
for x in range(10):
    if x == 15:
        break
    print(x)
else:
    print('Thanks')
```

```python
for x in range(10):
    if x == 5:
        continue
    print(x)
else:
    print('Thanks')
```

```python
while(True):
    marks = int(input('Enter marks: '))
    if marks < 0:
        break
else:
    print('Thanks')
```

```python
s = True
while(s):
    marks = int(input('Enter marks: '))
    if marks < 0:
        s = False
else:
    print('Thanks')
```
