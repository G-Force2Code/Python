#### 5. WAP to reverse a string taken as input from the user.

```python
strvalue = input("Enter a String which need to be reversed")
ResStr =""
for i in range(-1,-abs(len(strvalue)+1),-1):
    ResStr+=strvalue[i];   
print(ResStr)
```
